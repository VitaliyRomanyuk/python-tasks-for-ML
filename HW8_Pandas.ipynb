**4. Описова статистика**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній розмір рахунку на всьому наборі даних
- Максимальні чайові
- Кількість пропущених значень у всіх колонках
- Скільки в наборі даних рахунків за обіди і за вечері

```python
average_bill_size = df["total_bill"].mean()
print(f"Середній розмір рахунку: {average_bill_size:.2f}")
```

```python
max_tip = df["tip"].max()
print(f"Максимальні чайові, сума: {max_tip:.2f}")
```

```python
missing_values = df.isnull().sum()
print(f"Кількість пропущених значень у всіх колонках: {missing_values.sum()}")
```

```python
lunch_count = (df["time"] == "Lunch").sum()
dinner_count = (df["time"] == "Dinner").sum()

print(f"Обіди: {lunch_count}")
print(f"Вечері: {dinner_count}")
print(f"Разом: {lunch_count + dinner_count}")
```

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python
df.groupby("sex")["total_bill"].mean().round(2) # Середній рахунок за статтю
```

```python
df.groupby("day")["tip"].mean().round(2) # Середні чайові за днем тижня
```

```python
df.groupby("day").size() # Скільки відвідувачів у кожен день
```

**6. Створення нової колонки**

Створіть нову колонку `tip_percentage` - відсоток чайових від загальної суми рахунку.

**7. Сортування**

Відсортуйте датасет і виведіть результат сортування в кожному випадку окремо (кожен пункт оцінюється в 0.5 бала):
- За розміром рахунку (зростання)
- За чайовими (спадання)

Додатково виведіть топ-5 найбільших чайових використовуючи метод `nlargest`.

```python
sorted_by_bill = df.sort_values("total_bill") # Сортування за розміром рахунку (зростання)
print(sorted_by_bill.head())
```

```python
sorted_by_tip = df.sort_values("tip", ascending=False) # Сортування за чайовими (спадання)
print(sorted_by_tip.head()) 
```

```python
top5_tips = df.nlargest(5, "tip") # Топ-5 найбільших чайових через nlargest
print(top5_tips)
```
