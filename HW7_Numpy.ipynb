**2. Базові операції**

Створіть Numpy-масив зі значеннями [12, 24, 36, 48, 60].

Виконайте наступні операції над **оригінальним** масивом і запишіть у нову змінну результат (кожен пункт оцінюється в 0.5 бала):
- Помножте всі елементи на 2
- Додайте до всіх елементів 10
- Піднесіть всі елементи до квадрату
- Обчисліть квадратний корінь з усіх елементів

Виведіть результат виконання кожної операції.

```python
arr1 = np.array([12, 24, 36, 48, 60])
print(arr1)
arr2 = arr1 * 2 # Multiply each element by 2
print(arr2)
arr3 = arr1 + 10 # Add 10 to each element
print(arr3)
arr4 = arr1 ** 2 # Raise each element to the power of 2
print(arr4)
arr5 = np.sqrt(arr1).round(2) # Calculate the square root of each element and round to 2 decimal places
print(arr5)
```

**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

Обчисліть використовуючи функціонал numpy (кожен пункт оцінюється в 0.5 бала):
- Середнє значення
- Мінімум
- Максимум
- Суму всіх елементів

```python
data = np.array([18, 27, 33, 41, 52, 68, 79])
print(np.mean(data))    # Середнє значення
print(np.min(data))     # Мінімум
print(np.max(data))     # Максимум
print(np.sum(data))     # Сума всіх елементів
```

**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3 включно
- Елементи більші за 40

```python
arr6 = np.array([15, 22, 35, 44, 58, 67])
print(arr6[0])  # Перший елемент
print(arr6[-1]) # Останній елемент
print(arr6[1:4]) # Елементи з індексами 1 до 3 включно
print(arr6[arr6 > 40]) # Елементи більші за 40
```

**5. Двовимірні масиви**

Задано матрицю 3x3.

Виконайте наступні дії з функціоналом Numpy (кожен пункт оцінюється в 0.5 бала):
- Виведіть форму (shape) матриці
- Виведіть перший рядок
- Виведіть другу колонку
- Обчисліть суму всіх елементів
- Обчисліть середнє всіх елементів

```python
matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])
```

```python
print(matrix.shape)  # Форма (shape) матриці
print(matrix[0, :])  # Перший рядок
print(matrix[:, 1])  # Друга колонка
print(matrix.sum())  # Сума всіх елементів
print(matrix.mean())  # Cереднє всіх елементів
```
